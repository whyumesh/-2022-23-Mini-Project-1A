```python
# @title
from langchain_fireworks import ChatFireworks
from langchain.prompts import PromptTemplate
import os
```

```python
# @title
from google.colab import userdata
fireworks_api_key = userdata.get('fireworks_api_key')
```

```python
print(f"Fireworks API key: {fireworks_api_key}")
```

```python
# @title
os.environ["FIREWORKS_API_KEY"] = fireworks_api_key
```

```python
# @title
llm = ChatFireworks(
    model="accounts/fireworks/models/deepseek-v3",
    temperature=0,
    max_retries=2,
)
```

```python
# @title
output = llm.invoke("How many r's are there in strawberry?")
print(output)
```

```python
output.content
```

```python
# @title
template_string = '''
    You are an AI helpfull assistant.
    Please answer the following question.
    Question: {input}
    '''
```

```python
# @title
prompt = PromptTemplate(
        template=template_string,
        input_variables=['input'],
)
```

```python
# @title
query = "How many r's are there in strawberry"
```

```python
# @title
_input = prompt.format_prompt(input=query)
generated_query = llm.invoke(_input.to_string())
```
